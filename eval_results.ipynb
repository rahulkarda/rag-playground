# RAG Evaluation Results Notebook

This notebook summarizes answer relevance and faithfulness scores from experiments.

* Load output from eval_cli
* Show aggregate statistics
* Visualize distributions
* Show examples of high/low scoring answers
* TODO: Add more detailed analysis, config comparisons


```python
import json
import pandas as pd
import matplotlib.pyplot as plt

# Path to eval output file (update as needed)
output_path = 'eval_output.jsonl'

# Load results
results = []
with open(output_path, 'r', encoding='utf-8') as f:
    for line in f:
        if line.strip():
            results.append(json.loads(line))

df = pd.DataFrame(results)
print(f'Loaded {len(df)} results.')
```

```python
# Show basic statistics
df.describe(include='all')
```

```python
# Plot relevance score histogram
plt.hist(df['relevance_score'], bins=20, color='blue', alpha=0.7)
plt.xlabel('Relevance score')
plt.ylabel('Frequency')
plt.title('Answer relevance score distribution')
plt.show()
```

```python
# Plot faithfulness score histogram (if present)
if 'faithfulness_score' in df.columns:
    plt.hist(df['faithfulness_score'], bins=20, color='green', alpha=0.7)
    plt.xlabel('Faithfulness score')
    plt.ylabel('Frequency')
    plt.title('Answer faithfulness score distribution')
    plt.show()
```

```python
# Show examples of lowest/highest relevance
print('Lowest relevance:')
print(df.nsmallest(3, 'relevance_score')[['question', 'answer', 'ground_truth', 'relevance_score']])

print('
Highest relevance:')
print(df.nlargest(3, 'relevance_score')[['question', 'answer', 'ground_truth', 'relevance_score']])
```

```python
# Show examples of lowest/highest faithfulness (if present)
if 'faithfulness_score' in df.columns:
    print('Lowest faithfulness:')
    print(df.nsmallest(3, 'faithfulness_score')[['question', 'answer', 'ground_truth', 'faithfulness_score']])
    print('
Highest faithfulness:')
    print(df.nlargest(3, 'faithfulness_score')[['question', 'answer', 'ground_truth', 'faithfulness_score']])
```
